```python
import mysql.connector
import pandas as pd
```

```python
import getpass
pwd = getpass.getpass("MySQL Password:")

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password=pwd,
    database="data_ai"
)
print("Succesfull install mysql")
```

```python
def run_query(query):
    df = pd.read_sql(query, conn)
    df.index = range(1, len(df) + 1)
    return df
```

```python
# 1 Top record are their present in Table
query1="""Select count(*) Total from prai;"""
run_query(query1)
```

```python
# 2 Top Ai_tools use user for generat income
query2="""with a as (select user_type,ai_tool,monthly_income,
dense_rank() over (partition by user_type order by monthly_income desc) as rnk from prai)
select user_type,ai_tool,sum(monthly_income) as Total_income ,dense_rank() over (partition by user_type order by sum(monthly_income) desc) as
 rnk from a group by  user_type,ai_tool;"""
run_query(query2)
```

```python
# 3 Count How many propessionals and student productvity score more than average score
query3="""select user_type,count(*) as total ,round(avg(productivity_score),2) as Avg_Productvity from prai 
 group by user_type having avg(productivity_score)> (select round(avg(productivity_score),2) from prai);"""
run_query(query3)  
```

```python
# 4 Number of total  user recommendation of Use ai Would_recommend wise
Query5="""with a as(Select user_type,Would_Recommend,count(*) as Total from prai 
group by user_type,Would_Recommend order by user_type desc), 
b as(select user_type,Would_recommend,Total,count(*) over(partition by Would_Recommend) as Recommend 
from a order by Would_recommend)
select User_type,would_recommend ,total ,sum(Total) over(partition by Would_recommend) as Num_Recommend from b"""
run_query(Query5)
```

```python
#5 top 10 customer recommendation by Daily ai usages wise
Q="""select User_id,Gender,Country,User_Type,Education_Level,Age,AI_Tool,AI_Purpose,sum(AI_Usage_Hours_Per_Day) as Daily_Usages_hour,Would_recommend from prai
  group by User_id,Gender,Country,User_Type,Education_Level,Age,AI_Tool,AI_Purpose,Would_recommend order by Daily_Usages_hour desc limit 10;"""
run_query(Q)
```

```python
# 6 total Monthly income by user_type
Query6="""Select user_type,sum(Monthly_income) as total_Monthly_income from prai group by user_type order by sum(Monthly_income) desc; """
run_query(Query6)
```

```python
# 7 Top 10 customer 
Query7="""select user_id ,Ai_tool,Ai_purpose ,Would_recommend,sum(Monthly_income) as total_Monthly_income from prai
 group by user_id ,Ai_tool,Ai_purpose,Would_recommend order by sum(Monthly_income) desc limit 10;"""
run_query(Query7)
```

```python
# 8 count total recommend partition by would recommend education wise
Q8="""select education_level,would_recommend,count(*)as Total_reco from prai group by education_level,would_recommend order by Would_recommend desc,count(*) desc;"""
run_query(Q8)
```
