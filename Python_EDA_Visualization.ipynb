# Visualization

```python
plt.figure(figsize=(8,5))

plt.plot(df["AI_Usage_Hours_Per_Day"])

plt.title("AI Usage Trend")
plt.xlabel("Users")
plt.ylabel("AI Usage Hours")

plt.show()
```

```python
plt.figure(figsize=(8,5))

df["AI_Tool"].value_counts().plot(kind="bar")

plt.title("Most Popular AI Tools")
plt.xlabel("AI Tools")
plt.ylabel("Number of Users")

plt.show()
```

```python
plt.figure(figsize=(8,5))

df["AI_Purpose"].value_counts().plot(kind="barh")

plt.title("AI Usage by Purpose")
plt.xlabel("Number of Users")
plt.ylabel("AI Purpose")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Monthly_AI_Cost"].dropna(), bins=5)

plt.title("Monthly AI Cost Distribution")
plt.xlabel("Monthly AI Cost")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(7,7))

df["User_Type"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Students vs Professionals")

plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["AI_Usage_Hours_Per_Day"],
    df["Productivity_Score"]
)

plt.title("AI Usage vs Productivity")
plt.xlabel("AI Usage Hours")
plt.ylabel("Productivity Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Time_Saved_Hours_Per_Week"],bins=10)

plt.title(" Distribuition of Weekly Time Saved Through AI")
plt.xlabel("Hours Saved")
plt.ylabel("Users")

plt.show()
```
